```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.colors as mcolors
from scipy.stats import pearsonr

# Define the list of metrics to plot
metrics = ['tailbeat_frequency', 'standardized_amplitude', 'normalized_nnd', 'strouhal_number']
subplot_labels = ['(a)', '(b)', '(c)', '(d)']  # Labels for subplots

# Generate base colors using seaborn's "tab10" palette
tab10_palette = sns.color_palette("tab10", n_colors=10)

# Manually assign colors based on the requested order
metric_colors = {
    'normalized_nnd': tab10_palette[1],  
    'tailbeat_frequency': tab10_palette[2],  
    'standardized_amplitude': tab10_palette[3],  
    'strouhal_number': tab10_palette[4],  
}

# Function to create a darker version of a color
def darken_color(color, factor=0.7):
    rgba = mcolors.to_rgba(color)
    return (rgba[0] * factor, rgba[1] * factor, rgba[2] * factor)

# Function to get the scatter color and trendline color for each metric
def get_metric_palette(metric):
    base_color = metric_colors[metric]
    return base_color, darken_color(base_color, factor=0.7)

# Function to calculate tracklet averages for the metrics
def calculate_tracklet_averages(data, metrics):
    averages = []
    
    # Group by 'Trial' and 'tracklet' and calculate mean for each metric
    for (trial, tracklet), tracklet_data in data.groupby(['Trial', 'tracklet']):
        avg_values = {metric: tracklet_data[metric].mean() for metric in metrics}
        
        # Special case for normalized_nnd (only include values where unique_id_count > 1)
        if 'normalized_nnd' in metrics:
            valid_data = tracklet_data[tracklet_data['unique_id_count'] > 1]
            avg_values['normalized_nnd'] = valid_data['normalized_nnd'].mean() if not valid_data.empty else None
        
        avg_values['Trial'] = trial
        avg_values['Tracklet'] = tracklet
        avg_values['Standardized_Speed'] = tracklet_data['standardized_speed'].mean()
        averages.append(avg_values)
    
    return pd.DataFrame(averages)

# Calculate the averages for the metrics at the tracklet level
avg_df = calculate_tracklet_averages(df_filtered, metrics)

# Create a 2x2 grid layout for scatter plots
fig, axes = plt.subplots(2, 2, figsize=(14, 12))  # Increased figure size
axes = axes.flatten()

# Iterate over metrics and compute correlation + t-test
for i, metric in enumerate(metrics):
    scatter_color, trendline_color = get_metric_palette(metric)
    
    # Ensure only matching data is used for correlation
    plot_df = avg_df.dropna(subset=[metric, 'Standardized_Speed'])
    
    # Scatterplot
    sns.scatterplot(data=plot_df, x='Standardized_Speed', y=metric, color=scatter_color, ax=axes[i], s=80)  # Larger markers
    
    # Trendline
    sns.regplot(data=plot_df, x='Standardized_Speed', y=metric, scatter=False, 
                line_kws={'color': trendline_color, 'lw': 3}, ax=axes[i])  # Thicker trendline

    # Compute Pearson correlation if there are at least 2 valid data points
    if len(plot_df) > 1:
        pearson_r, pearson_p = pearsonr(plot_df['Standardized_Speed'], plot_df[metric])
        t_value = (pearson_r * (len(plot_df) - 2) ** 0.5) / (1 - pearson_r**2) ** 0.5 if abs(pearson_r) < 1 else float('inf')
        dof = len(plot_df) - 2
    else:
        pearson_r, pearson_p, t_value, dof = float('nan'), float('nan'), float('nan'), 0  # Not enough data
    
    # Print results
    print(f"Metric: {metric}")
    print(f" Pearson Correlation: R = {pearson_r:.3f}, p-value = {pearson_p:.3g}")
    print(f" t-value = {t_value:.3f}, dof = {dof}")
    print("-" * 50)

    # Display correlation and t-test result on plot
    axes[i].text(0.05, 0.85, f'R = {pearson_r:.2f}', 
                 transform=axes[i].transAxes, fontsize=14, color='black', 
                 bbox=dict(facecolor='white', alpha=0.5))  # Increased font size
    
    # Add subplot label (e.g., (a), (b), etc.)
    axes[i].text(-0.1, 1.1, subplot_labels[i], transform=axes[i].transAxes, 
                 fontsize=16, fontweight='bold', va='top', ha='right')  # Bigger and bold subplot labels

   # Set labels and title with bigger font size
    axes[i].set_title(f'Speed vs {metric_title.get(metric, metric)}', fontsize=16)
    axes[i].set_xlabel('Speed (BL/s)', fontsize=14)
    axes[i].set_ylabel(metric_titles.get(metric, metric), fontsize=14)

    
    # Increase tick label sizes
    axes[i].tick_params(axis='both', which='major', labelsize=12)

# Adjust layout and show the plot
plt.tight_layout()
plt.savefig(f'{figures_path}metrics_scatterplot.png')
plt.show()

```

### Carangid boxplots

```python
# Define the list of metrics
metrics = ['standardized_speed', 'normalized_nnd', 'tailbeat_frequency', 
           'standardized_amplitude', 'strouhal_number']

# Function to calculate tracklet averages for the metrics
def calculate_tracklet_averages(data, metrics):
    averages = []
    
    for (trial, tracklet, carangid), tracklet_data in data.groupby(['Trial', 'tracklet', 'carangids']):
        avg_values = {metric: tracklet_data[metric].mean() for metric in metrics}  # Compute mean for metrics
        avg_values['Trial'] = trial  # Include trial info
        avg_values['Tracklet'] = tracklet  # Include tracklet info
        avg_values['Carangid'] = carangid  # Include carangid info
        averages.append(avg_values)  
    
    return pd.DataFrame(averages)

# Calculate the averages for the metrics at the tracklet level
avg_df = calculate_tracklet_averages(df, metrics)

# Ensure 'Carangid' is treated as a categorical variable
avg_df['Carangid'] = avg_df['Carangid'].astype('category')

# ✅ Apply filtering: Only keep values < 2 for the specified metrics
avg_df.loc[avg_df['strouhal_number'] >= 2, 'strouhal_number'] = None

# Drop NaN values to avoid plotting empty values
avg_df_filtered = avg_df.dropna(subset=['strouhal_number'], how='all')
avg_df_filtered = avg_df_filtered[avg_df_filtered['normalized_nnd'] >= 0]


def bootstrap_mean_diff(data, metric, group_col='Carangid', n_bootstrap=1000, ci=95):
    # Extract values for each group
    group_0 = data[data[group_col] == 0][metric].dropna().values
    group_1 = data[data[group_col] == 1][metric].dropna().values

    # Observed difference in means
    obs_diff = np.mean(group_0) - np.mean(group_1)

    # Bootstrap resampling
    boot_diffs = np.array([
        np.mean(np.random.choice(group_0, size=len(group_0), replace=True)) - 
        np.mean(np.random.choice(group_1, size=len(group_1), replace=True))
        for _ in range(n_bootstrap)
    ])

    # Compute confidence interval
    lower = np.percentile(boot_diffs, (100 - ci) / 2)
    upper = np.percentile(boot_diffs, 100 - (100 - ci) / 2)

    # Compute two-tailed p-value
    extreme_count = np.sum(np.abs(boot_diffs) >= np.abs(obs_diff))
    p_value = extreme_count / n_bootstrap  # Proper normalization

    # Ensure p-value does not exceed 1
    p_value = min(p_value, 1)

    return obs_diff, (lower, upper), p_value, boot_diffs

# Store results
results = []
bootstrap_distributions = {}

for metric in metrics:
    obs_diff, conf_int, p_val, boot_dist = bootstrap_mean_diff(avg_df_filtered, metric)
    
    results.append({
        'Metric': metric,
        'Observed Difference': obs_diff,
        '95% CI': conf_int,
        'p-value': p_val
    })

    # Store bootstrap distributions for plotting if needed
    bootstrap_distributions[metric] = boot_dist

# Convert results to DataFrame
results_df = pd.DataFrame(results)
print(results_df)
```

```python
import numpy as np
import pandas as pd

# Function to calculate tracklet averages for the metrics
def calculate_tracklet_averages(data, metrics):
    averages = []
    
    for (trial, tracklet, carangid), tracklet_data in data.groupby(['Trial', 'tracklet', 'carangids']):
        avg_values = {metric: tracklet_data[metric].mean() for metric in metrics}  # Compute mean for metrics
        avg_values['Trial'] = trial  # Include trial info
        avg_values['Tracklet'] = tracklet  # Include tracklet info
        avg_values['Carangid'] = carangid  # Include carangid info
        averages.append(avg_values)  
    
    return pd.DataFrame(averages)

# Calculate the averages for the metrics at the tracklet level
avg_df = calculate_tracklet_averages(df, metrics)



def bootstrap_p_value_2sample(data, metric, group_col='Carangid', B=1000, ci=95):
    """
    Compute the p-value under the bootstrap method for two independent groups

    Parameters
    ----------
    data : DataFrame
        The dataset containing the values
    metric : str
        The column name for the metric being tested
    group_col : str
        The column name indicating group membership
    B : int, optional
        Number of bootstrap samples (default is 10,000)
    ci : int, optional
        Confidence interval percentage (default is 95%)

    Returns
    -------
    obs_diff : float
        The observed difference in means
    conf_int : tuple
        The confidence interval for the bootstrap distribution
    p_value : float
        The p-value from the bootstrap test
    boot_diffs : array
        The bootstrap distribution of mean differences
    """

    # Extract values for each group
    group_0 = data[data[group_col] == 0][metric].dropna().values
    group_1 = data[data[group_col] == 1][metric].dropna().values
    
    print(group_0.shape,group_1.shape)
    # Compute the observed difference in means
    obs_diff = np.nanmean(group_0) - np.nanmean(group_1)

    # Combine both groups into a single array
    # combined = np.concatenate([group_0, group_1])

    # Bootstrap resampling
    boot_diffs = np.zeros(B)
    count = 0
    for b in range(B):
        boot_group0 = np.random.choice(group_0,size=len(group_0),replace = True)
        boot_group1 = np.random.choice(group_1,size=len(group_1),replace = True)
        
        boots_difference = boot_group0.mean() - boot_group1.mean()
        
        # boot_sample = np.random.choice(combined, size=len(combined), replace=True)
        # boot_diffs[b] = np.mean(boot_sample[:len(group_0)]) - np.mean(boot_sample[len(group_0):])
        boot_diffs[b] = boots_difference
    # Compute two-tailed p-value
    #p_value = np.mean(np.abs(boot_diffs) >= np.abs(obs_diff))
    for el in boot_diffs:
        if abs(el) >= abs(obs_diff):
            count+=1
    p_value = count / B
    print(p_value)
    # Compute confidence interval
    lower = np.percentile(boot_diffs, 2.5)
    upper = np.percentile(boot_diffs, 97.5)
    conf_int = (lower, upper)

    return obs_diff, conf_int, p_value, boot_diffs

# Store results
boot_results = []
bootstrap_distributions = {}

for metric in metrics:
    obs_diff, conf_int, p_val, boot_dist = bootstrap_p_value_2sample(avg_df, metric)
    
    boot_results.append({
        'Metric': metric,
        'Observed Difference': obs_diff,
        '95% CI': conf_int,
        'p-value': p_val
    })

    # Store bootstrap distributions for plotting
    bootstrap_distributions[metric] = boot_dist

# Convert results to DataFrame
boot_results_df = pd.DataFrame(boot_results)
print(boot_results_df)

```

```python
def calculate_p_values(df, metric):
    """Performs Wilcoxon Signed-Rank test for paired data or Mann-Whitney U test for unpaired data."""
    # Group by Trial and Tracklet, and pivot data for Carangid == 0 and Carangid == 1
    paired_data = df[df['Metric'] == metric].pivot_table(
        index=['Trial', 'Tracklet'], 
        columns='Carangid', 
        values='Value', 
        aggfunc='mean'
    )

    # Remove rows where either Carangid == 0 or Carangid == 1 is missing
    paired_data = paired_data.dropna(subset=[0, 1])

    # If enough paired data for Wilcoxon Signed-Rank Test
    if len(paired_data) > 2:
        group_0 = paired_data[0]
        group_1 = paired_data[1]
        n_pairs = len(group_0)  # Number of paired data points
        if n_pairs > 2:
            stat, p_value = wilcoxon(group_0, group_1)
            print(f"Wilcoxon test performed for {metric}: p={p_value}, n={n_pairs}")
            return p_value, n_pairs, n_pairs, 'Wilcoxon'  # Ensure 4 return values
    else:
        # If not enough paired data, perform Mann-Whitney U Test instead (unpaired test)
        group_0 = df[(df['Metric'] == metric) & (df['Carangid'] == 0)]['Value']
        group_1 = df[(df['Metric'] == metric) & (df['Carangid'] == 1)]['Value']
        n_group_0 = len(group_0)
        n_group_1 = len(group_1)
        
        if n_group_0 > 2 and n_group_1 > 2:  # Ensure there are enough points
            stat, p_value = mannwhitneyu(group_0, group_1, alternative="two-sided")
            print(f"Mann-Whitney test performed for {metric}: p={p_value}, n0={n_group_0}, n1={n_group_1}")
            return p_value, n_group_0, n_group_1, 'Mann-Whitney'

    print(f"Not enough data for {metric} to perform a test")
    return None, 0, 0, None  # Ensure 4 return values, even if no test is performed

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import wilcoxon, mannwhitneyu
import numpy as np

# Define the list of metrics
metrics = ['standardized_speed', 'normalized_nnd', 'tailbeat_frequency', 
           'standardized_amplitude', 'strouhal_number']

# Function to calculate tracklet averages for the metrics
def calculate_tracklet_averages(data, metrics):
    averages = []
    
    for (trial, tracklet, carangid), tracklet_data in data.groupby(['Trial', 'tracklet', 'carangids']):
        avg_values = {metric: tracklet_data[metric].mean() for metric in metrics}
        avg_values['Trial'] = trial
        avg_values['Tracklet'] = tracklet
        avg_values['Carangid'] = carangid
        averages.append(avg_values)
    
    return pd.DataFrame(averages)

# Calculate averages at tracklet level
avg_df = calculate_tracklet_averages(df, metrics)

# Ensure 'Carangid' is categorical
avg_df['Carangid'] = avg_df['Carangid'].astype('category')

# ✅ Filter values: Only keep values < 2 for 'strouhal_number'
avg_df.loc[avg_df['strouhal_number'] >= 2, 'strouhal_number'] = None

# Drop NaN values
avg_df_filtered = avg_df.dropna(subset=['strouhal_number'], how='all')

# Pivot to check for paired data
paired_df = avg_df_filtered.pivot(index=['Trial', 'Tracklet'], columns='Carangid', values=metrics)

# Function to choose Wilcoxon or Mann-Whitney and compute RBC
def perform_test(paired_df, metric):
    """Runs Wilcoxon if paired, otherwise Mann-Whitney U, and calculates correct RBC."""
    if metric in paired_df.columns.levels[0]:
        try:
            group_0 = paired_df[(metric, 0)].dropna()
            group_1 = paired_df[(metric, 1)].dropna()
            
            # Check if fully paired
            paired_mask = group_0.index.intersection(group_1.index)
            if len(paired_mask) > 2:
                # Wilcoxon Signed-Rank Test
                w_stat, p_value = wilcoxon(group_0.loc[paired_mask], group_1.loc[paired_mask])
                
                # Correct Rank-Biserial Correlation for Wilcoxon
                n = len(paired_mask)
                 # Corrected Rank-Biserial Correlation (RBC)
                rbc = 1 - (2 * w_stat) / (n * (n + 1)) if n > 0 else None

                test_used = "Wilcoxon Signed-Rank"
            else:
                # Mann-Whitney U Test
                u_stat, p_value = mannwhitneyu(group_0, group_1, alternative="two-sided")

                # Correct Rank-Biserial Correlation for Mann-Whitney U
                n0, n1 = len(group_0), len(group_1)
                rbc = 1 - (2 * u_stat) / (n0 * n1)

                test_used = "Mann-Whitney U"
            
            return {
                "Metric": metric, 
                "Test Used": test_used,
                "p-value": p_value, 
                "Effect Size (RBC)": rbc
            }
        except ValueError:
            return {"Metric": metric, "Error": "Test failed (possibly all values are identical)"}
    else:
        return {"Metric": metric, "Error": "Metric not found"}

# Run tests for all metrics
results = [perform_test(paired_df, metric) for metric in metrics]

# Convert results to a DataFrame
results_df = pd.DataFrame(results)
print(results_df)

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import GammaRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.utils import resample

# 1. Prepare the Data
y = nndTracklet["standardized_speed"].values.reshape(-1, 1).astype(np.float32)
nndTracklet["log_groupsize"] = np.log(nndTracklet["unique_id_count"])

scaler = StandardScaler()
nndTracklet["log_groupsize"] = scaler.fit_transform(nndTracklet[["log_groupsize"]])

# Create the interaction term
nndTracklet["interaction"] = nndTracklet["carangids"] * nndTracklet["log_groupsize"]

# Select features, including interaction
X = nndTracklet[["carangids", "log_groupsize", "interaction"]]
y = y.ravel()

# Fit the Gamma Regression Model
clf = GammaRegressor(alpha=0)
clf.fit(X, y)

# 2. Create Predictions for New Data
group_sizes_orig = np.linspace(1, 4, 100)  # Group sizes from 1 to 4
log_group_sizes = np.log(group_sizes_orig)  # Log-transform
group_sizes = scaler.transform(log_group_sizes.reshape(-1, 1)).flatten()  # Standardize

# Design Matrices for Both Treatments
treatment0 = np.zeros_like(group_sizes)
interaction0 = group_sizes * treatment0  # All zeros
X_treatment0 = np.column_stack([treatment0, group_sizes, interaction0])

treatment1 = np.ones_like(group_sizes)
interaction1 = group_sizes * treatment1  # Equals group_sizes
X_treatment1 = np.column_stack([treatment1, group_sizes, interaction1])

# Convert to DataFrame with correct column names
X_treatment0 = pd.DataFrame(np.column_stack([treatment0, group_sizes, interaction0]), columns=X.columns)
X_treatment1 = pd.DataFrame(np.column_stack([treatment1, group_sizes, interaction1]), columns=X.columns)

# Now predictions will work without warnings
predictions_treatment0 = clf.predict(X_treatment0)
predictions_treatment1 = clf.predict(X_treatment1)


# 3. Bootstrap Confidence Intervals
n_bootstraps = 1000  # Number of bootstrap samples
percentile = 95  # Confidence interval level

bootstrap_preds0 = np.zeros((n_bootstraps, len(group_sizes)))
bootstrap_preds1 = np.zeros((n_bootstraps, len(group_sizes)))

for i in range(n_bootstraps):
    X_resampled, y_resampled = resample(X, y, random_state=i)  # Resample data
    clf_bootstrap = GammaRegressor(alpha=0)
    clf_bootstrap.fit(X_resampled, y_resampled)
    
    bootstrap_preds0[i, :] = clf_bootstrap.predict(X_treatment0)
    bootstrap_preds1[i, :] = clf_bootstrap.predict(X_treatment1)

# Compute Confidence Intervals
lower_bound0, upper_bound0 = np.percentile(bootstrap_preds0, [(100 - percentile) / 2, 100 - (100 - percentile) / 2], axis=0)
lower_bound1, upper_bound1 = np.percentile(bootstrap_preds1, [(100 - percentile) / 2, 100 - (100 - percentile) / 2], axis=0)

# 4. Plot Results with Confidence Intervals
plt.figure(figsize=(8, 6))
plt.plot(group_sizes_orig, predictions_treatment0, label="Treatment 0", color="blue")
plt.fill_between(group_sizes_orig, lower_bound0, upper_bound0, color="blue", alpha=0.2, label="95% CI (Treatment 0)")

plt.plot(group_sizes_orig, predictions_treatment1, label="Treatment 1", color="red")
plt.fill_between(group_sizes_orig, lower_bound1, upper_bound1, color="red", alpha=0.2, label="95% CI (Treatment 1)")

plt.xlabel("Group Size")
plt.ylabel("Predicted Speed")
plt.title("Predicted Speed by Group Size and Treatment with 95% CI")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import GammaRegressor

# Dictionary to store models, coefficients, and confidence intervals
models = {}
coefficients = {}
ci_lower = {}
ci_upper = {}

# Number of bootstrap samples for confidence intervals
n_bootstrap = 1000
percentile_low, percentile_high = 2.5, 97.5  # 95% confidence interval

# Loop through each metric and fit a GammaRegressor
for metric in metrics:
    # Filter dataset for nearest neighbor distance (exclude group size 1)
    if metric == "normalized_nnd":
        filtered_data = nndTracklet[nndTracklet["unique_id_count"] > 1].copy()
    else:
        filtered_data = nndTracklet.copy()


    # Log-transform group size
    filtered_data["log_groupsize"] = np.log(filtered_data["unique_id_count"])

    # Standardize log-transformed group size
    scaler = StandardScaler()
    filtered_data["log_groupsize"] = scaler.fit_transform(filtered_data[["log_groupsize"]])

    # Create interaction term
    filtered_data["interaction"] = filtered_data["carangids"] * filtered_data["log_groupsize"]

    # Drop rows where the target metric has NaN values
    filtered_data = filtered_data.dropna(subset=[metric, "carangids", "log_groupsize", "interaction"])
    
    # Select features
    X = filtered_data[["carangids", "log_groupsize", "interaction"]]
    y = filtered_data[metric].values.reshape(-1, 1).astype(np.float32).ravel()

    # Fit the model
    clf = GammaRegressor(alpha=0)
    clf.fit(X, y)

    models[metric] = clf  # Store model
    coefficients[metric] = clf.coef_  # Store coefficients

    # Define group size range (exclude 1 if needed)
    if metric == "normalized_nnd":
        group_sizes_orig = np.linspace(2, 4, 100)  # Exclude 1
    else:
        group_sizes_orig = np.linspace(1, 4, 100)

    # Log-transform group sizes
    log_group_sizes = np.log(group_sizes_orig)

    # Standardize using the fitted scaler
    group_sizes_scaled = scaler.transform(log_group_sizes.reshape(-1, 1)).flatten()

    # Create predictor arrays for each treatment group
    treatment0 = np.zeros_like(group_sizes_scaled)
    interaction0 = group_sizes_scaled * treatment0  # All zeros

    treatment1 = np.ones_like(group_sizes_scaled)
    interaction1 = group_sizes_scaled * treatment1  # Equals group_sizes_scaled

    # Build the design matrices
    X_treatment0 = pd.DataFrame(np.column_stack([treatment0, group_sizes_scaled, interaction0]), columns=X.columns)
    X_treatment1 = pd.DataFrame(np.column_stack([treatment1, group_sizes_scaled, interaction1]), columns=X.columns)
    
    # Get predictions using the fitted model
    predictions_treatment0 = clf.predict(X_treatment0)
    predictions_treatment1 = clf.predict(X_treatment1)

    # Bootstrap to estimate confidence intervals
    boot_predictions_treatment0 = np.zeros((n_bootstrap, len(group_sizes_orig)))
    boot_predictions_treatment1 = np.zeros((n_bootstrap, len(group_sizes_orig)))

    for i in range(n_bootstrap):
        # Resample from the filtered dataset (ensuring group size 1 is excluded when needed)
        sample_indices = np.random.choice(len(X), size=len(X), replace=True)
        X_sample, y_sample = X.iloc[sample_indices], y[sample_indices]

        clf_boot = GammaRegressor(alpha=0)
        clf_boot.fit(X_sample, y_sample)

        boot_predictions_treatment0[i, :] = clf_boot.predict(X_treatment0)
        boot_predictions_treatment1[i, :] = clf_boot.predict(X_treatment1)

    # Compute 95% confidence intervals
    ci_lower[metric] = {
        "treatment0": np.percentile(boot_predictions_treatment0, percentile_low, axis=0),
        "treatment1": np.percentile(boot_predictions_treatment1, percentile_low, axis=0)
    }
    ci_upper[metric] = {
        "treatment0": np.percentile(boot_predictions_treatment0, percentile_high, axis=0),
        "treatment1": np.percentile(boot_predictions_treatment1, percentile_high, axis=0)
    }

    print(f"Gamma Regression Score for {metric}: {clf.score(X, y)}")
    print(f"Coefficients for {metric}: {clf.coef_}\n")

    # Store predictions for later use in plotting
    models[metric] = {
        "group_sizes_orig": group_sizes_orig,
        "predictions_treatment0": predictions_treatment0,
        "predictions_treatment1": predictions_treatment1,
        "ci_lower": ci_lower[metric],
        "ci_upper": ci_upper[metric],
    }

    
   
```

```python
# Log-transform groupsize
nndTracklet["log_groupsize"] = np.log(nndTracklet["unique_id_count"])

scaler = StandardScaler()
nndTracklet['log_groupsize'] = scaler.fit_transform(nndTracklet[['log_groupsize']])

# Create interaction term
nndTracklet["interaction"] = nndTracklet["carangids"] * nndTracklet["log_groupsize"]

# Select features
X = nndTracklet[["carangids", "log_groupsize", "interaction"]]

# Dictionary to store models and coefficients
models = {}
coefficients = {}

# Loop through each metric and fit a GammaRegressor
for metric in metrics:
    y = nndTracklet[metric].values.reshape(-1, 1).astype(np.float32)
    y = y.ravel()  # Flatten array
    

    clf = linear_model.GammaRegressor(alpha = 0)
    clf.fit(X, y)
    clf.score(X, y)
    
    models[metric] = clf  # Store model
    coefficients[metric] = clf.coef_  # Store coefficients

    # 1. Create a range of original group size values (from 1 to 5).
    group_sizes = np.linspace(1, 4, 100)
    
    ## 2. Log-transform the group sizes.
    log_group_sizes = np.log(group_sizes)
    
    # # 3. Standardize the log-transformed group sizes.
    # # In practice, use the scaler fitted on your training data.
    scaler = StandardScaler()
    group_sizes_scaled = scaler.fit_transform(log_group_sizes.reshape(-1, 1)).flatten()
    
    # 4. Create predictor arrays for each treatment group.
    # For treatment 0:
    treatment0 = np.zeros_like(group_sizes_scaled)
    interaction0 = group_sizes_scaled * treatment0  # This will be all zeros
    
    # For treatment 1:
    treatment1 = np.ones_like(group_sizes_scaled)
    interaction1 = group_sizes_scaled * treatment1  # This equals group_sizes_scaled

    # 5. Build the design matrices.
    # The order of columns must match the order used during training.
    X_treatment0 = np.column_stack([group_sizes, treatment0, interaction0])
    X_treatment1 = np.column_stack([group_sizes, treatment1, interaction1])
    
    # Get predictions using the fitted model
    predictions_treatment0 = clf.predict(X_treatment0)
    predictions_treatment1 = clf.predict(X_treatment1)

    print(f"Gamma Regression Score for {metric}: {clf.score(X, y)}")
    print(f"Coefficients for {metric}: {clf.coef_}\n")

    # Plot the predictions
    plt.figure(figsize=(8, 6))
    plt.plot(group_sizes[:len(predictions_treatment0)], predictions_treatment0, label="Treatment 0")
    plt.plot(group_sizes[:len(predictions_treatment1)], predictions_treatment1, label="Treatment 1")
    plt.xlabel("Group Size")
    plt.ylabel(f"Predicted {metric}")
    plt.title(f"Predicted {metric} by Group Size and Treatment")
    plt.legend()
    plt.show()
```

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import numpy as np
from scipy.stats import linregress

# Dictionary to store model results
results = {}

# Function to check linearity trend across unique_id_count
def check_trend(df_subset):
    """
    Uses linear regression to assess whether the response variable follows a smooth trend.
    If the R² is high, we assume a smooth trend and use Polynomial Coding.
    Otherwise, we use Sum Coding.
    """
    # Group by unique_id_count and take mean to observe trend
    trend_df = df_subset.groupby("unique_id_count")["Value"].mean().reset_index()

    # Perform a simple linear regression
    slope, intercept, r_value, p_value, std_err = linregress(trend_df["unique_id_count"], trend_df["Value"])

    # If R² is high (say > 0.8), use Polynomial; otherwise, use Sum
    return "Poly" if r_value**2 > 0.8 else "Sum"

# Loop through each metric and fit the best Gamma regression model
for metric in nndTracklet_melted["Metric"].unique():
    df_subset = nndTracklet_melted[nndTracklet_melted["Metric"] == metric].dropna(subset=["Value"])  # Remove NaNs
    
    if df_subset.empty:  # Skip if the subset is empty after filtering
        print(f"Skipping {metric} due to no valid data.")
        continue
    
    # Convert 'unique_id_count' to categorical and remove unused categories for this metric
    df_subset["unique_id_count"] = df_subset["unique_id_count"].astype("category")
    df_subset["unique_id_count"] = df_subset["unique_id_count"].cat.remove_unused_categories()

    # Decide on Sum or Polynomial Coding based on trend
    contrast_type = check_trend(df_subset)
    
    # Construct formula dynamically based on chosen contrast
    formula = f"Value ~ carangids * C(unique_id_count, {contrast_type})"

    # Fit the Gamma regression model
    model = smf.glm(formula, 
                    data=df_subset, 
                    family=sm.families.Gamma(link=sm.families.links.log())).fit()
    
    # Store model summary
    results[metric] = model.summary()

# Print results for each metric
for metric, summary in results.items():
    print(f"\n==== {metric} (Using {check_trend(nndTracklet_melted[nndTracklet_melted['Metric'] == metric])} Coding) ====\n")
    print(summary)
```

#### Transition 0->1

```python
# ------------------------------------------------
# time_range defines the number of frames before and after the transition.
# ------------------------------------------------
time_range = 300

# List to store each transition window DataFrame
transition_windows = []

# Iterate over each trial.
for trial, trial_data in df.groupby('Trial'):
    # For each trial, first identify all transition events (for each individual).
    # We’ll store each event as a tuple: (center, source_id)
    trial_transitions = []
    
    for id_val, id_data in trial_data.groupby('ID'):
        # Ensure the data are sorted by the existing 'frame' column.
        id_data_sorted = id_data.sort_values('frame').reset_index(drop=True)
        # Compute the difference in carangids.
        diff_vals = id_data_sorted['carangids'].diff()
        # Find positions where the difference equals 1.
        trans_positions = diff_vals[diff_vals == 1].index.tolist()
        
        for pos in trans_positions:
            # Only consider events where the individual has at least time_range frames
            # before and after the event.
            if pos >= time_range and pos + time_range < len(id_data_sorted):
                center = id_data_sorted.loc[pos, 'frame']
                trial_transitions.append((center, id_val))
    
    # For each transition event detected in this trial, extract the corresponding window
    # for every individual in the trial.
    for center, source_id in trial_transitions:
        start_frame = center - time_range
        end_frame = center + time_range
        
        # Define the complete range of relative frames we expect: from -time_range to +time_range.
        rel_index = list(range(-time_range, time_range + 1))
        
        # For every individual in the trial, extract the window.
        for id_val, id_data in trial_data.groupby('ID'):
            # Subset the data for frames within the window.
            subset = id_data[(id_data['frame'] >= start_frame) & (id_data['frame'] <= end_frame)].copy()
            # Compute the relative frame for each row.
            subset['relative_frame'] = subset['frame'] - center
            
            # Reindex on the relative_frame so that the complete window is present.
            # Any missing frames will be filled with NaN.
            subset = subset.set_index('relative_frame').reindex(rel_index)
            
            # Add meta-information about the trial and transition.
            subset['Trial'] = trial
            subset['transition_center'] = center
            subset['source_id'] = source_id
            subset['ID'] = id_val
            
            # Reset index so that relative_frame is a column.
            subset = subset.reset_index()
            
            # Append this window DataFrame to our list.
            transition_windows.append(subset)

# Combine all the individual transition windows into one DataFrame.
transitions_df = pd.concat(transition_windows, ignore_index=True)

# transitions_df now contains, for each trial and each detected transition event,
# a complete window (from -time_range to +time_range relative to the event) for every ID.
# Missing frames (if an individual doesn't have enough frames) are filled with NaN.
```

```python
# ------------------------------------------------
# time_range defines the number of frames before and after the transition.
# ------------------------------------------------
time_range = 300

# List to store each transition window DataFrame
transition_windows = []

# Iterate over each trial.
for trial, trial_data in df.groupby('Trial'):
    # For each trial, first identify all transition events (for each individual).
    # We’ll store each event as a tuple: (center, source_id)
    trial_transitions = []
    
    for id_val, id_data in trial_data.groupby('ID'):
        # Ensure the data are sorted by the existing 'frame' column.
        id_data_sorted = id_data.sort_values('frame').reset_index(drop=True)
        # Compute the difference in carangids.
        diff_vals = id_data_sorted['carangids'].diff()
        # Find positions where the difference equals 1.
        trans_positions = diff_vals[diff_vals == -1].index.tolist()
        
        for pos in trans_positions:
            # Only consider events where the individual has at least time_range frames
            # before and after the event.
            if pos >= time_range and pos + time_range < len(id_data_sorted):
                center = id_data_sorted.loc[pos, 'frame']
                trial_transitions.append((center, id_val))
    
    # For each transition event detected in this trial, extract the corresponding window
    # for every individual in the trial.
    for center, source_id in trial_transitions:
        start_frame = center - time_range
        end_frame = center + time_range
        
        # Define the complete range of relative frames we expect: from -time_range to +time_range.
        rel_index = list(range(-time_range, time_range + 1))
        
        # For every individual in the trial, extract the window.
        for id_val, id_data in trial_data.groupby('ID'):
            # Subset the data for frames within the window.
            subset = id_data[(id_data['frame'] >= start_frame) & (id_data['frame'] <= end_frame)].copy()
            # Compute the relative frame for each row.
            subset['relative_frame'] = subset['frame'] - center
            
            # Reindex on the relative_frame so that the complete window is present.
            # Any missing frames will be filled with NaN.
            subset = subset.set_index('relative_frame').reindex(rel_index)
            
            # Add meta-information about the trial and transition.
            subset['Trial'] = trial
            subset['transition_center'] = center
            subset['source_id'] = source_id
            subset['ID'] = id_val
            
            # Reset index so that relative_frame is a column.
            subset = subset.reset_index()
            
            # Append this window DataFrame to our list.
            transition_windows.append(subset)

# Combine all the individual transition windows into one DataFrame.
leavingtransitions_df = pd.concat(transition_windows, ignore_index=True)
```

```python
import numpy as np
import pandas as pd

# Define the list of metrics you want to calculate the mean for
metrics = ['standardized_speed', 'normalized_nnd', 'tailbeat_frequency', 
           'standardized_amplitude', 'strouhal_number']

# Group by 'unique_id_count', 'Trial_num', and 'tracklet', then calculate mean for each metric
nndTracklet = df[['unique_id_count', 'Trial_num', 'tracklet'] + metrics].groupby(
    ['unique_id_count', 'Trial_num', 'tracklet']
).mean().reset_index()

nndTracklet = nndTracklet[
    (nndTracklet["unique_id_count"] >= 1) & 
    (nndTracklet["unique_id_count"] <= 4)
]


def bootstrap_anova_p_value(data, metric, group_col='unique_id_count', B=10000, ci=95):
    """
    Perform a bootstrapped one-way ANOVA test.

    Parameters
    ----------
    data : DataFrame
        The dataset containing the values.
    metric : str
        The column name for the metric being tested.
    group_col : str, optional
        The column name indicating group membership (default is 'group').
    B : int, optional
        Number of bootstrap samples (default is 10,000).
    ci : int, optional
        Confidence interval percentage (default is 95%).

    Returns
    -------
    obs_F : float
        The observed F-statistic.
    conf_int : tuple
        The confidence interval for the bootstrap F-distribution.
    p_value : float
        The bootstrapped p-value.
    boot_Fs : array
        The bootstrap distribution of F-statistics.
    """


    # Group data
    groups = data[group_col].unique()
    group_values = {g: data[data[group_col] == g][metric].dropna().values for g in groups}

    # Compute overall mean
    overall_mean = np.mean(data[metric].dropna())

    # Compute observed F-statistic
    SS_between = sum(len(vals) * (np.mean(vals) - overall_mean) ** 2 for vals in group_values.values())
    SS_within = sum(sum((vals - np.mean(vals)) ** 2) for vals in group_values.values())

    df_between = len(groups) - 1
    df_within = len(data) - len(groups)

    obs_F = (SS_between / df_between) / (SS_within / df_within)

    # Bootstrap Resampling
    boot_Fs = np.zeros(B)
    combined = data[metric].dropna().values

    for b in range(B):
        # Shuffle values while keeping group sizes fixed
        np.random.shuffle(combined)
        shuffled_groups = {g: combined[i:i+len(group_values[g])] for i, g in enumerate(group_values)}

        # Compute F-statistic for bootstrap sample
        SS_between_boot = sum(len(vals) * (np.mean(vals) - np.mean(combined)) ** 2 for vals in shuffled_groups.values())
        SS_within_boot = sum(sum((vals - np.mean(vals)) ** 2) for vals in shuffled_groups.values())

        boot_Fs[b] = (SS_between_boot / df_between) / (SS_within_boot / df_within)

    # Compute p-value (proportion of bootstrapped F's greater than observed F)
    p_value = np.mean(boot_Fs >= obs_F)

    # Compute confidence interval
    lower = np.percentile(boot_Fs, (100 - ci) / 2)
    upper = np.percentile(boot_Fs, 100 - (100 - ci) / 2)
    conf_int = (lower, upper)

    return obs_F, conf_int, p_value, boot_Fs

# Store results
anova_results = []
bootstrap_distributions = {}

# Loop through each metric and perform bootstrapped ANOVA
for metric in metrics:
    # Filter out NaNs for the current metric
    filtered_df = nndTracklet.dropna(subset=[metric])

    # If the metric is 'nnd_normalised', also filter out unique_id_count == 1
    if metric == 'nnd_normalised':
        filtered_df = filtered_df[filtered_df['unique_id_count'] != 1]

    # Perform bootstrapped ANOVA
    obs_F, conf_int, p_value, boot_Fs = bootstrap_anova_p_value(filtered_df, metric, group_col='unique_id_count')
    
    # Store results
    anova_results.append({
        'Metric': metric,
        'Observed F': obs_F,
        '95% CI': conf_int,
        'p-value': p_value
    })

    # Store bootstrap distributions for plotting
    bootstrap_distributions[metric] = boot_Fs

# Convert results to DataFrame
anova_results_df = pd.DataFrame(anova_results)
print(anova_results_df)
```

```python
from itertools import combinations
from statsmodels.stats.multitest import multipletests

def bootstrap_pairwise_p_values(data, metric, group_col, B=10000):
    """
    Perform bootstrapped pairwise comparisons for multiple groups.

    Parameters
    ----------
    data : DataFrame
        Filtered dataset containing the metric and group column.
    metric : str
        The column name for the metric being tested.
    group_col : str
        The column name indicating group membership.
    B : int, optional
        Number of bootstrap samples (default is 10,000).

    Returns
    -------
    pairwise_results : DataFrame
        A DataFrame containing pairwise p-values and confidence intervals.
    """

    # Get unique groups
    groups = data[group_col].unique()
    group_combinations = list(combinations(groups, 2))  # All pairwise combinations
    results = []

    for g1, g2 in group_combinations:
        # Extract values for the two groups
        vals1 = data[data[group_col] == g1][metric].dropna().values
        vals2 = data[data[group_col] == g2][metric].dropna().values

        # Compute observed difference in means
        obs_diff = np.mean(vals1) - np.mean(vals2)

        # Combine both groups into a single array
        combined = np.concatenate([vals1, vals2])

        # Bootstrap resampling
        boot_diffs = np.zeros(B)
        for b in range(B):
            boot_sample = np.random.choice(combined, size=len(combined), replace=True)
            boot_diffs[b] = np.mean(boot_sample[:len(vals1)]) - np.mean(boot_sample[len(vals1):])

        # Compute p-value
        p_value = np.mean(np.abs(boot_diffs) >= np.abs(obs_diff))

        # Compute confidence intervals
        lower = np.percentile(boot_diffs, 2.5)
        upper = np.percentile(boot_diffs, 97.5)
        conf_int = (lower, upper)

        results.append({
            'Group 1': g1,
            'Group 2': g2,
            'Observed Difference': obs_diff,
            '95% CI': conf_int,
            'p-value': p_value
        })

    # Convert to DataFrame
    pairwise_results_df = pd.DataFrame(results)

    # Adjust p-values for multiple comparisons (Bonferroni correction)
    pairwise_results_df['Adjusted p-value'] = multipletests(pairwise_results_df['p-value'], method='bonferroni')[1]

    return pairwise_results_df

# Main analysis
anova_results = []
pairwise_comparisons = {}

for metric in metrics:
    # Filter NaNs
    filtered_df = nndTracklet.dropna(subset=[metric])

    # If metric is 'nnd_normalised', remove unique_id_count == 1
    if metric == 'nnd_normalised':
        filtered_df = filtered_df[filtered_df['unique_id_count'] != 1]

    # Perform bootstrapped ANOVA
    obs_F, conf_int, p_value, boot_Fs = bootstrap_anova_p_value(filtered_df, metric, group_col='unique_id_count')
    
    # Store ANOVA results
    anova_results.append({
        'Metric': metric,
        'Observed F': obs_F,
        '95% CI': conf_int,
        'p-value': p_value
    })

    # If ANOVA is significant, perform pairwise tests
    if p_value < 0.05:
        pairwise_comparisons[metric] = bootstrap_pairwise_p_values(filtered_df, metric, group_col='unique_id_count')

# Convert ANOVA results to DataFrame
anova_results_df = pd.DataFrame(anova_results)
print(anova_results_df)

# Print pairwise comparisons if available
for metric, df in pairwise_comparisons.items():
    print(f"\nPairwise comparisons for {metric}:\n", df)
```

```python
from scipy.stats import f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

anova_results = {}

for metric in metrics:
    # Exclude unique_id_count = 1 for NND
    if metric == "normalized_nnd":
        filtered_df = nndTracklet[nndTracklet["unique_id_count"] > 1]
    else:
        filtered_df = nndTracklet
    
    # Get unique groups
    unique_groups = filtered_df["unique_id_count"].unique()

    # Create a list of data groups
    groups = [filtered_df[filtered_df["unique_id_count"] == group][metric].dropna() 
              for group in unique_groups]
    
    # Perform ANOVA test
    stat, p_value = f_oneway(*groups)

    # Compute variance, mean, and standard deviation for each group
    group_stats = {
        group: {
            "mean": np.mean(filtered_df[filtered_df["unique_id_count"] == group][metric].dropna()),
            "variance": np.var(filtered_df[filtered_df["unique_id_count"] == group][metric].dropna(), ddof=1),
            "std_dev": np.std(filtered_df[filtered_df["unique_id_count"] == group][metric].dropna(), ddof=1)
        }
        for group in unique_groups
    }

    # Store sample size and degrees of freedom
    n_total = sum(len(g) for g in groups)  # Total sample size
    df_between = len(unique_groups) - 1  # Degrees of freedom between groups
    df_within = n_total - len(unique_groups)  # Degrees of freedom within groups

    # Store results
    anova_results[metric] = {
        "N": n_total,
        "df_between": df_between,
        "df_within": df_within,
        "F-statistic": stat,
        "p-value": p_value,
        "group_stats": group_stats,
        "post_hoc": None  # Placeholder for post-hoc results
    }

    # If ANOVA is significant, perform Tukey's HSD test
    if p_value < 0.05:
        tukey_data = filtered_df[[metric, "unique_id_count"]].dropna()
        tukey = pairwise_tukeyhsd(endog=tukey_data[metric], groups=tukey_data["unique_id_count"], alpha=0.05)
        anova_results[metric]["post_hoc"] = tukey.summary()

# Convert results to DataFrame
anova_df = pd.DataFrame.from_dict(anova_results, orient="index")

# Print results in a formatted way
print("\n=== ANOVA Test Results ===")
for metric, values in anova_results.items():
    # Convert p-value to scientific notation if too small
    p_value_formatted = f"{values['p-value']:.5f}" if values["p-value"] >= 0.0001 else f"{values['p-value']:.2e}"

    print(f"\nMetric: {metric}")
    print(f"  N = {values['N']}, df_between = {values['df_between']}, df_within = {values['df_within']}")
    print(f"  F = {values['F-statistic']:.3f}, p = {p_value_formatted}")
    print("  Group Statistics:")
    for group, stats in values["group_stats"].items():
        print(f"    Group {group}: Mean = {stats['mean']:.5f}, Variance = {stats['variance']:.5f}, Std Dev = {stats['std_dev']:.5f}")
   # Print post-hoc results if ANOVA was significant
    if values["p-value"] < 0.05:
        print("\n  Post-hoc Tukey HSD Test:")
        
        # Convert Tukey's HSD summary to a DataFrame for better formatting
        post_hoc_df = pd.DataFrame(data=values["post_hoc"].data[1:], columns=values["post_hoc"].data[0])
        
        # Format p-values in the post-hoc test results
        def format_p_value(p):
            if p < 1.0e-5:  # Threshold to avoid 0.00e+00
                return "< 1.0e-5"
            return f"{p:.5f}" if p >= 0.0001 else f"{p:.2e}"
        
        post_hoc_df["p-adj"] = post_hoc_df["p-adj"].apply(format_p_value)
        
        print(post_hoc_df)
```

```python
def assign_group_letters(p_matrix, groups):
    """
    Assigns compact letter display (CLD) labels to indicate statistical significance.
    Groups that are not significantly different share the same letter.
    """
    # Initialize group labels
    group_labels = {g: set() for g in groups}
    letters = "abcdefghijklmnopqrstuvwxyz"  # Allow for many groups

    # Start assigning letters
    current_letter = 0
    for (i, group1), (j, group2) in itertools.combinations(enumerate(groups), 2):
        p_value = p_matrix.iloc[i, j]

        if p_value < 0.05:  # Significant difference
            if not group_labels[group1]:
                group_labels[group1].add(letters[current_letter])
            if not group_labels[group2]:
                current_letter += 1  # Move to next letter
                group_labels[group2].add(letters[current_letter])
        else:
            # If groups are not significantly different, share a letter
            common_letter = list(group_labels[group1] | group_labels[group2])
            if common_letter:
                group_labels[group1].update(common_letter)
                group_labels[group2].update(common_letter)
            else:
                group_labels[group1].add(letters[current_letter])
                group_labels[group2].add(letters[current_letter])

    # Convert sets to sorted strings
    return {g: "".join(sorted(letters)) for g, letters in group_labels.items()}

# Compute Dunn’s test results and assign group letters
dunn_labels = {}

for metric in metrics:
    # Exclude unique_id_count = 1 for NND
    if metric == "normalized_nnd":
        filtered_df = nndTracklet[nndTracklet["unique_id_count"] > 1]
    else:
        filtered_df = nndTracklet
    
    # Perform Dunn’s test
    dunn_test = posthoc_dunn(filtered_df, val_col=metric, group_col="unique_id_count", p_adjust="bonferroni")
    
    # Assign significance letters
    groups = sorted(filtered_df["unique_id_count"].unique())  # Ensure sorted order
    dunn_labels[metric] = assign_group_letters(dunn_test, groups)

print("\n=== Dunn’s Test Group Labels ===")
for metric, labels in dunn_labels.items():
    print(f"\nMetric: {metric}")
    for group, label in labels.items():
        print(f"  Group {group}: {label}")
```

```python
dunn_results = {}

for metric in metrics:
    # Exclude unique_id_count = 1 for NND
    if metric == "normalized_nnd":
        filtered_df = nndTracklet[nndTracklet["unique_id_count"] > 1]
    else:
        filtered_df = nndTracklet
    
    # Perform Dunn’s test with Bonferroni correction
    dunn_test = posthoc_dunn(filtered_df, val_col=metric, group_col="unique_id_count", p_adjust="bonferroni")
    
    # Store results
    dunn_results[metric] = dunn_test

    # Print formatted results
    print(f"\n=== Dunn’s Test Results for {metric} ===")
    for i, group1 in enumerate(dunn_test.index):
        for j, group2 in enumerate(dunn_test.columns):
            if i < j:  # Only print upper triangle (avoid duplicates)
                z_value = (dunn_test.iloc[i, j] - 0.5) * 2  # Approximate Z-score
                p_value = dunn_test.iloc[i, j]
                print(f"  {group1} vs {group2}: Z = {z_value:.3f}, p = {p_value:.5f}")
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Define the list of metrics
metrics = ['standardized_speed', 'normalized_nnd', 'tailbeat_frequency', 'standardized_amplitude', 'strouhal_number']

# Group by 'unique_id_count', 'Trial_num', and 'tracklet', then calculate mean for each metric
nndTracklet = df[['unique_id_count', 'Trial_num', 'tracklet'] + metrics].groupby(
    ['unique_id_count', 'Trial_num', 'tracklet']
).mean().reset_index()

# Melt dataframe for plotting
nndTracklet_melted = nndTracklet.melt(id_vars=['unique_id_count', 'Trial_num', 'tracklet'],
                                      value_vars=metrics,
                                      var_name='Metric',
                                      value_name='Value')

# Function to compute pairwise comparisons using LMM
def lmm_pairwise(data, metric):
    model = smf.mixedlm("Value ~ C(unique_id_count)", data, groups=data["tracklet"]).fit()
    comparisons = model.t_test_pairwise("C(unique_id_count)").result_frame
    comparisons["sig"] = comparisons["P>|t|"] < 0.05  # Mark significant results
    return comparisons

# Function to assign letter groupings
def assign_groups(pairwise_results):
    groups = {}
    sorted_groups = sorted(pairwise_results.index.levels[0])
    
    for i, grp1 in enumerate(sorted_groups):
        grp_label = set()
        for j, grp2 in enumerate(sorted_groups):
            if i == j or pairwise_results.loc[(grp1, grp2), "P>|t|"] >= 0.05:
                grp_label.add(grp2)
        groups[grp1] = "".join(sorted(grp_label))
    return groups

# Create subplots
n_metrics = len(metrics)
n_cols = 3
n_rows = -(-n_metrics // n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 5))
axes = axes.flatten()

# Iterate through metrics
for i, metric in enumerate(metrics):
    ax = axes[i]
    metric_data = nndTracklet_melted[nndTracklet_melted['Metric'] == metric]
    
    # Boxplot
    sns.boxplot(x='unique_id_count', y='Value', data=metric_data, ax=ax)
    sns.stripplot(x='unique_id_count', y='Value', data=metric_data, alpha=0.6, jitter=True, ax=ax)
    
    # Pairwise comparisons with LMM
    try:
        pairwise_results = lmm_pairwise(metric_data, metric)
        group_assignments = assign_groups(pairwise_results)
        
        # Annotate p-values
        for j, group in enumerate(metric_data['unique_id_count'].unique()):
            ax.text(j, metric_data['Value'].max(), group_assignments[group], ha='center', fontsize=12, fontweight='bold')
            p_value = pairwise_results.loc[(group, metric_data['unique_id_count'].unique()[0]), "P>|t|"]
            if p_value < 0.05:
                ax.text(j, metric_data['Value'].max() * 0.95, f"* (p={p_value:.3f})", ha='center', fontsize=10, color='red')
    except:
        print(f"LMM failed for {metric}")
    
    ax.set_title(metric)
    ax.set_xlabel("Unique ID Count")
    ax.set_ylabel(metric)

# Adjust layout and show plot
plt.tight_layout()
plt.show()
```
